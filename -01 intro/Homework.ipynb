## Median value of horsepower

```python
#Calculate the initial median
initial_median = df['horsepower'].median()
print(f"Initial Median: {initial_median}")

#Calculate the most frequent value (Mode)
most_frequent = df['horsepower'].mode()[0]
print(f"Most Frequent Value (Mode): {most_frequent}")

#Fill missing values with the mode and re-calculate median
df_filled = df['horsepower'].fillna(most_frequent)
new_median = df_filled.median()

print(f"New Median: {new_median}")
#Check if it changed
print(f"Has it changed? {'Yes' if initial_median != new_median else 'No'}")
```

## Sum of weights

```python
#Isolate the first 7 Asian cars and select weight/year columns
asia_cars = df[df['origin'] == 'Asia'][['vehicle_weight', 'model_year']].head(7)

#Convert to underlying NumPy array X
X = asia_cars.values

#Compute X^T * X
XTX = X.T.dot(X)

#Invert XTX
XTX_inv = np.linalg.inv(XTX)

#Define target vector y
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

#Multiply the inverse with X^T, then multiply by y to get weights w
w = XTX_inv.dot(X.T).dot(y)
print(f"Calculated Weights (w): {w}")

#Get the sum of all components in w
total_sum = w.sum()
print(f"Sum of all elements: {total_sum:.4f}")
```
